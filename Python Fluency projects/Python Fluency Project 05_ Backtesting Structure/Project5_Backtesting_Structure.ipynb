### 10. Long and short trade monitoring

````python
# After entry block, already in trade 

for i in range(len(df)):

    # Store details for monitoring

    # NOT IN TRADE BLOCK
    if not in_trade:
        # not in trade, so, look for entry signal
        continue

    # ENTRY BLOCK
    if not pd.isna(signal) and i + 1 < len(df):
        # entry block
        entry_px = df["Ratio"].iloc[i + 1]
        signal_type = signal
        peak = entry_px
        trough = entry_px
        
    # TRADE MONITORING BLOCK
    elif in_trade:
        if signal_type == "reverting from support":

            # LONG trade
            peak = max(peak, price)
            trough = min(trough, price)

            drawdown_pct = (price - peak) / peak
            dd_mvol = drawdown_pct / mvol

        elif signal_type == "reverting from resistance":

            # SHORT trade
            peak = min(peak, price)
            trough = max(trough, price)

            drawdown_pct = -1 * ((price - peak) / peak)
            dd_mvol = drawdown_pct / mvol
````

````python
trades = []

in_trade = False

entry_px = None
entry_date = None
signal_type = None

for i in range(len(df)):
    price = df["Ratio"].iloc[i]
    date = df.index[i]
    signal = df["final_signal"].iloc[i]
    mvol = df["mvol"].iloc[i]

    if not in_trade:


        # look for signal
        if not pd.isna(signal) and i+1 < len(df):

            # ENTRY BLOCK
            entry_px = df["Ratio"].iloc[i+1]
            entry_date = df.index[i+1]

            signal_type = signal

            peak = entry_px
            trough = entry_px

            in_trade = True

            entry_index = i+1

    elif in_trade:

        # calc dd, peak, trough
        if signal_type == "reverting from support":
            peak = max(peak, price)
            trough = min(trough, price)

            drawdown_pct = price / peak - 1

        elif signal_type == "reverting from resistance":
            peak = min(peak, price)
            trough = max(trough, price)

            drawdown_pct = -1* (price / peak - 1)


        # we check for exit and do some calcs

        # exit condition
        dd_mvol = drawdown_pct / mvol

        if dd_mvol < -0.6   and i + 1 < len(df):                    # boundary check, in case the last row doesn't exist
            exit_date = df.index[i+1]
            exit_px = df["Ratio"].iloc[i+1]

            return_perc = exit_px / entry_px - 1
            mfe = peak / entry_px - 1
            mae = trough / entry_px - 1

            if signal_type == "reverting from support":

                directional_return = return_perc
                directional_mae = mae
                directional_mfe = mfe

            elif signal_type == "reverting from resistance":

                directional_return = -return_perc
                directional_mae = -mae
                directional_mfe = -mfe

            trades.append({
                "entry_date": entry_date,
                "entry_px": entry_px,
                "signal_type": signal_type,
                "dd_mvol": dd_mvol,
                "returns": directional_return,
                "mfe": directional_mfe,
                "mae": directional_mae,
                "exit_date": exit_date,
                "exit_px": exit_px,
                "holding_period": (i+1) - entry_index

            })

            # reset variables
            in_trade = False

            peak = None
            trough = None

            signal_type = None
            # It is overwritten at the next entry anyway, but resetting it keeps your trade state clean.

            # resetting other variables: It's a defensive programming practice that prevents future bugs.
            entry_index = None
            entry_date = None
            entry_px = None
            exit_px = None
            exit_date = None



trades_df = pd.DataFrame(trades)
````

# Calculations

````python
# 1. Number of trades
num_trades = len(trades_df)
````

````python
# 2. Win rate
winning_trades = (trades_df["returns"] > 0).sum()

win_rate = (
    (winning_trades / num_trades) * 100
    if num_trades > 0 else np.nan
)
````

````python
# 3. Expected value (average return per trade)
expected_value = trades_df["returns"].mean()
````

````python
# 4. Volatility of trade returns
volatility = trades_df["returns"].std()
````

````python
# 5. payoff ratio

avg_win = trades_df[trades_df["returns"] > 0]["returns"].mean()
avg_loss = trades_df[trades_df["returns"] < 0]["returns"].mean()

payoff_ratio = avg_win / avg_loss if pd.notna(avg_loss) and avg_loss != 0 else np.nan
payoff_ratio
````

````python
# 6. Maximum loss
max_loss = trades_df["returns"].min()
````

````python
performance_list = []

performance_list.append({
    "num_trades": num_trades,
    "win_rate%": win_rate,
    "ev%": expected_value*100,
    "volatility%": volatility*100,
    "payoff_ratio": payoff_ratio,
    "max_loss": max_loss
})

performance = pd.DataFrame(performance_list)
````
